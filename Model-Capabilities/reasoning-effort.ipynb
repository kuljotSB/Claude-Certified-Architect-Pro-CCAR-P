### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

claude_model_name = os.getenv("CLAUDE_MODEL_NAME")
claude_api_key = os.getenv("CLAUDE_API_KEY")
```

### Creating the Anthropic Client and using the Messages API

```python
import anthropic

client = anthropic.Anthropic(api_key=claude_api_key)

user_prompt = """ You are a senior digital marketing consultant.

A client has an online electronics store.

Current Metrics:
- Bounce Rate: 71%
- Conversion Rate: 1.8%
- Average Session Duration: 42 seconds
- Mobile Traffic: 78%

Suggest practical ways to improve conversions.
Explain why each recommendation would help.
Prioritize recommendations from highest to lowest impact. """

print(f"\n\n### Reasoning Effort: {"medium".capitalize()}\n")

message = client.messages.create(
        model=claude_model_name,
        system = "You are a helpful AI assistant",
        messages=[
            {
                "role": "user",
                "content": user_prompt,
            }
        ],
        output_config = {
            "effort": "medium"
        },
        max_tokens = 10000
)

for block in message.content:
        if block.type == "text":
            print(block.text)

print("\n usage stats:", message.usage)

print("\n\n" + "-" * 80 + "\n\n")
```
